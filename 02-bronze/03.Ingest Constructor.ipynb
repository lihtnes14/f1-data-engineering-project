```python
source_file = f"{landing_folder_path}/constructors.json"
table_name = f"{catlog_name}.{bronze_schema}.constructors"
```

###Step 1 - Read the JSON file using the dataframe reader API

```python
#Define the schema
constructor_schema = """constructorId STRING,
                        name STRING, 
                        nationality STRING, 
                        url STRING
                    """
```

```python
constructor_df = (
    spark.read
    .format('json')
    .schema(constructor_schema)
    .option('mode', 'FAILFAST')
    .load(source_file)
)
```

```python
display(constructor_df)
```

```python
constructor_df_final = add_ingestion_metadata(constructor_df)
```

```python
display(constructor_df_final)
```

```python
constructor_df_final.write.mode('overwrite').format('delta').saveAsTable(table_name)
```
