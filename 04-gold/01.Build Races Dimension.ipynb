```python
from pyspark.sql import functions as F
```

```python
target_table = f"{catlog_name}.{gold_schema}.dim_races"
```

```python
target_table
```

```python
circuits_df = spark.table(f"{catlog_name}.{silver_schema}.circuits")
races_df = spark.table(f"{catlog_name}.{silver_schema}.races")
```

```python
dim_races_df = (
    races_df
        .join(
            circuits_df,
            races_df.circuit_id == circuits_df.circuit_id,
            "inner"
            )
        .select(
            races_df.season,
            races_df.round,
            races_df.race_name,
            races_df.race_date,
            circuits_df.circuit_name,
            circuits_df.locality,
            circuits_df.country,

        )
)
```

```python
display(dim_races_df)
```

```python
(
    dim_races_df
        .write
        .format("delta")
        .saveAsTable(target_table)
)
```
